````python
import requests
import pandas as pd
# limiting the fetched data to the first 50,000 rows that include latitude and longitude
NY_crashes = "https://data.cityofnewyork.us/resource/h9gi-nx95.json?$limit=50000&$where=latitude IS NOT NULL AND longitude IS NOT NULL"
# creating a data frame with columns for collision ID, number of persons killed, 
# number of persons injured, number of cyclists injured, latitude, and longitude
response = requests.get(NY_crashes)
data = response.json()
crash_df = pd.DataFrame(data)[
    ["number_of_persons_killed", "number_of_cyclist_killed", "latitude", "longitude"]
].copy()
crash_df = crash_df.apply(pd.to_numeric, errors="coerce")
crash_df.head()
````

````python
crash_df.mean()
````

````python
crash_df.std()
````

### Justification / description of attribute selection
* I included only fatalities of persons and cyclists to determine where the most lethal collisions tend to happen, but this approach does not take into account injuries from collisions that could lead to fatal complications or much worse quality of life afterwards. I chose to calculate descriptive statistics for latitude and longitude to determine where the most dangerous areas for collisions are, as this information could be the most useful in improving road safety measures.
* The average latitude and longitude are useful because they indicate the average location of the accidents (the data could only include deaths to determine the most lethal locations). Based on the results of ```crash_df.std()```,
there is less deviation in the average latitude (std = 3.166857) than average longitude (std = 5.746671), suggesting that the most dangerous areas are concentrated vertically rather than horizontally.
